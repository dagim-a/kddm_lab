```python
# 2. load the source tables into a raw schema
```

```python
RAW_TABLES: dict[str, dict[str, str]] = {
    "customers": {"customer_zip_code_prefix": "VARCHAR"},  # keep leading zeros
    "orders": {},
    "order_items": {},
    "products": {},
    "sellers": {}
}
```

```python
def csv_path(table: str) -> Path:
    """"""
    return RAW_DIR / f"olist_{table}_dataset.csv"

def check_source_files(tables) -> None:
    """Fail once, list every missing file"""
    missing = [csv_path(t) for t in tables if not csv_path(t).exists()]
    if missing:
        names = "\n".join(f" - {p}" for p in missing)
        raise FileNotFoundError(f"Missing CSV files: {names}")
```

```python
RAW_SCHEMA = "raw"
```

```python
def load_raw_table(con: duckdb.DuckDBPyConnection, table: str, column_types: dict[str, str]) -> int:
    target = f'{RAW_SCHEMA}."{table}"'
    source = con.read_csv(str(csv_path(table)), header=True, dtype=column_types or None)
    con.execute(f"DROP TABLE IF EXISTS {target}")
    source.to_table(target)
    return con.execute(f"SELECT count(*) FROM {target}").fetchone()[0]
```

```python
con.execute(f"CREATE SCHEMA IF NOT EXISTS {RAW_SCHEMA}")
```

```python
# 5.1 orders with no items 
```

```python
q("""
SELECT
    o.order_status,
    COUNT(*) AS orders_without_items
FROM raw.orders AS o
LEFT JOIN raw.order_items AS i
    ON o.order_id = i.order_id
WHERE i.order_id IS NULL
GROUP BY o.order_status
ORDER BY orders_without_items DESC
""")
```

```python
# 5.2 check order status and class imbalance 
```

```python
q("""
WITH status_counts AS (
    SELECT
        order_status,
        COUNT(*) AS n
    FROM raw.orders
    GROUP BY order_status
)
SELECT
    order_status,
    n,
    ROUND(100.0 * n / SUM(n) OVER (), 2) AS pct
FROM status_counts
ORDER BY n DESC
""")
```

```python
# 5.3 missing and inconsistent timestamps
```

```python
q("""
SELECT
    COUNT(*) FILTER (
        WHERE order_approved_at IS NULL
    ) AS approved_null,
    COUNT(*) FILTER (
        WHERE order_delivered_carrier_date IS NULL
    ) AS carrier_null,
    COUNT(*) FILTER (
        WHERE order_delivered_customer_date IS NULL
    ) AS delivered_null,
    COUNT(*) FILTER (
        WHERE order_status = 'delivered'
          AND order_delivered_customer_date IS NULL
    ) AS delivered_but_no_date,
    COUNT(*) FILTER (
        WHERE order_delivered_carrier_date > order_delivered_customer_date
    ) AS carrier_after_customer
FROM raw.orders
""")
```

```python
monthly = q("""
SELECT
    strftime(date_trunc('month', order_purchase_timestamp), '%Y-%m') AS month,
    COUNT(*) AS orders
FROM raw.orders
WHERE order_purchase_timestamp IS NOT NULL
GROUP BY 1
ORDER BY 1
""")
ax = monthly.plot.bar(x="month", y="orders", figsize=(16, 6), legend=False, color="#000000")
ax.set_title("Orders per month: partial months at both ends")
ax.set_xlabel("")
plt.tight_layout(); plt.show()
```

```python
# Prove the primary key
```

```python
q("""
WITH key_stats AS (
    SELECT
        'customers' AS "Table",
        'customer_id' AS "Key column(s)",
        COUNT(*) AS n_rows,
        COUNT(DISTINCT customer_id) AS n_distinct,
        COUNT(*) FILTER (WHERE customer_id IS NULL) AS n_nulls
    FROM raw.customers

    UNION ALL

    SELECT
        'orders',
        'order_id',
        COUNT(*),
        COUNT(DISTINCT order_id),
        COUNT(*) FILTER (WHERE order_id IS NULL)
    FROM raw.orders

    UNION ALL

    SELECT
        'products',
        'product_id',
        COUNT(*),
        COUNT(DISTINCT product_id),
        COUNT(*) FILTER (WHERE product_id IS NULL)
    FROM raw.products

    UNION ALL

    SELECT
        'sellers',
        'seller_id',
        COUNT(*),
        COUNT(DISTINCT seller_id),
        COUNT(*) FILTER (WHERE seller_id IS NULL)
    FROM raw.sellers

    UNION ALL

    SELECT
        'order_items',
        'order_id + order_item_id',
        COUNT(*),
        COUNT(DISTINCT (order_id, order_item_id)),
        COUNT(*) FILTER (
            WHERE order_id IS NULL OR order_item_id IS NULL
        )
    FROM raw.order_items
)
SELECT
    "Table",
    "Key column(s)",
    n_rows,
    n_distinct,
    n_nulls,
    n_nulls = 0 AND n_rows = n_distinct AS "Valid key?"
FROM key_stats
ORDER BY "Table";
""")
```

```python
# Measure the cardinality
```

```python
q("""
    SELECT 'customers -> orders' AS relationship,
       MIN(n) AS min_children, MAX(n) AS max_children
    FROM (SELECT c.customer_id, COUNT(o.order_id) AS n
      FROM raw.customers c LEFT JOIN raw.orders o ON o.customer_id = c.customer_id
      GROUP BY c.customer_id)
    UNION ALL
    SELECT 'orders -> order_items', MIN(n), MAX(n)
    FROM (SELECT o.order_id, COUNT(i.order_item_id) AS n
      FROM raw.orders o LEFT JOIN raw.order_items i ON i.order_id = o.order_id
      GROUP BY o.order_id)
    UNION ALL
    SELECT 'products -> order_items', MIN(n), MAX(n)
    FROM (SELECT p.product_id, COUNT(i.order_item_id) AS n
      FROM raw.products p LEFT JOIN raw.order_items i ON i.product_id = p.product_id
      GROUP BY p.product_id)
    UNION ALL
    SELECT 'sellers -> order_items', MIN(n), MAX(n)
    FROM (SELECT s.seller_id, COUNT(i.order_item_id) AS n
      FROM raw.sellers s LEFT JOIN raw.order_items i ON i.seller_id = s.seller_id
      GROUP BY s.seller_id);
""")
```

```python
# 4A, Two columns in the customers table look like customer identifiers. How do they
# differ, and which one would you use to count customers?
```

```python
# The customer_id is a unique identifier assigned to each order, so it may appear different even when the same person places multiple orders. In contrast, customer_unique_id identifies the actual customer and stays the same across their orders. To find the total number of unique customers, use COUNT(DISTINCT customer_unique_id). Counting customer_id would give you the total number of orders rather than the number of individual customers.
```

```python
# 4B, Which table is a junction table, and which relationships does it resolve?
```

```python
# The order_items table connects orders, products, and sellers. Since one order can contain multiple products and the same product can appear in different orders, this table helps manage that many-to-many relationship. Each row represents a specific item within an order, identified by order_id and order_item_id. It also links products to sellers through product_id and seller_id, making it easier to track which seller provides each product.
```
